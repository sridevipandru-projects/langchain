```python
!pip install -U langchain langchain-core langchain-google-genai
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.prompts import ChatPromptTemplate,SystemMessagePromptTemplate, HumanMessagePromptTemplate
gemini_Secret_Key=userdata.get('geminiSecretKey')
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", api_key=gemini_Secret_Key,temperature=0.6)
# response = llm.invoke("Give road map for programming language-java")
# print(response.content)
prompt = PromptTemplate(
    input_variables=["language"],
    template="""
Give me a roadmap for the programming language "{language}" in four bullet points.
The bullet points must include:
1. Prerequisites to learn the language
2. Basic concepts
3. Intermediate concepts
4. Resources required
"""
)
# promptresult=prompt.format(language="java")
# print(promptresult)
# chain = prompt | llm
# response = chain.invoke({"language": "java"})
# print(response.content)

system_message = SystemMessagePromptTemplate.from_template(
    "You are a translator asistant ")

human_message = HumanMessagePromptTemplate.from_template(
    "Translate '{sentence}' from english to spanish."
)
chat_prompt = ChatPromptTemplate.from_messages([system_message, human_message])
chain = chat_prompt | llm
response = chain.invoke({"sentence": "too many cooks spoil the broth"})
print(response.content)
```
